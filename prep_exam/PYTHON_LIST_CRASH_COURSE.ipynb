### Python list crashcourse

```python
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True]  # Can mix types
empty = []                         # Empty list

print(mixed)
print(mixed[1]) #2nd element
print(mixed[0]) #1st element
print(mixed[-1]) #last element
print(mixed[-2]) #2nd last element
print(mixed[1:3]) #slicing 2nd to 3rd element
print(mixed[:3]) #slicing start to 3rd element
print(mixed[2:]) #slicing 3rd element to end
print(mixed[:]) #slicing whole list
print(mixed[::2]) #slicing with step of 2
print(mixed[::-1]) #reversing list
print(mixed[::-2]) #reversing list with step of 2
print(mixed[::2][::-1]) #slicing with step of 2 and reversing
print(len(mixed)) #length of list
```

#### Modifying lists

```python
fruits = ["apple", "banana", "cherry"]
print(fruits)

# adding elements
fruits.append("orange") #add to end
fruits.insert(1, "blueberry") #add at index 1
fruits.extend(["kiwi", "mango"]) #add multiple elements
fruits += ["grape", "peach"] #add multiple elements using +=
fruits = fruits + ["plum", "pear"] #add multiple elements using +

fruits.remove("banana") #remove by value
poped = fruits.pop() #remove last element
print("Popped element:", poped)

fruits.pop(5) #remove by index
fruits.remove("kiwi") #remove by value

# ["grape", "peach"] remove by array of values
to_remove = ["grape", "peach"]
fruits = [fruit for fruit in fruits if fruit not in to_remove]

for fruit in ["mango", "plum"]:
    fruits.remove(fruit) #remove by value in loop


del fruits[2] #remove by index
print("After deletion:", fruits)

fruits *= 2 #duplicate list

print(fruits)
```
