# 02 - NLTK Corpus Operations
**MIDTERM: Gutenberg, Brown corpus, WordNet synonyms/antonyms**

```python
# === IMPORTS ===
import nltk
from nltk.corpus import gutenberg, brown, wordnet as wn
from nltk import FreqDist
```

---
## PATTERN 1: Gutenberg Corpus

```python
# List all available files
print(gutenberg.fileids())
```

```python
# Get all words from a file
words = gutenberg.words('austen-persuasion.txt')
print('Total words:', len(words))
print('First 20 words:', list(words[:20]))
```

```python
# Word types (unique words)
word_types = len(set(words))
print('Unique word types:', word_types)
```

```python
# Frequency distribution - most common word
fdist = FreqDist(words)
print('Most common words:', fdist.most_common(10))
print('Most common word:', fdist.max())
```

```python
# Lexical richness
richness = len(words) / len(set(words))
print('Lexical richness:', richness)
```

## PATTERN 2: Brown Corpus

```python
# List categories (genres)
print(brown.categories())
```

```python
# Get words from a category
news_words = brown.words(categories='news')
print('News words count:', len(news_words))
```

```python
# Get tagged sentences (for POS tagging)
tagged_sents = brown.tagged_sents(tagset='universal')
print('Number of tagged sentences:', len(tagged_sents))
print('First sentence:', tagged_sents[0])
```

```python
# Train/test split for taggers
size = int(len(tagged_sents) * 0.9)
train_sents = tagged_sents[:size]
test_sents = tagged_sents[size:]
print(f'Train: {len(train_sents)}, Test: {len(test_sents)}')
```

## PATTERN 3: WordNet - Synsets

```python
# Get all synsets for a word
synsets = wn.synsets('dog')
print('Synsets for "dog":', synsets)
```

```python
# Get definition of first synset
dog_synset = wn.synset('dog.n.01')
print('Definition:', dog_synset.definition())
```

```python
# Get examples
print('Examples:', dog_synset.examples())
```

---
## PATTERN 4: WordNet - Synonyms

```python
# Get synonyms (lemma names from synset)
synset = wn.synset('good.a.01')
lemmas = synset.lemmas()
synonyms = [lemma.name() for lemma in lemmas]
print('Synonyms of good:', synonyms)
```

```python
# Get ALL synonyms across ALL synsets
all_synonyms = []
for synset in wn.synsets('happy'):
    for lemma in synset.lemmas():
        all_synonyms.append(lemma.name())

print('All synonyms of happy:', set(all_synonyms))
```

---
## PATTERN 5: WordNet - Antonyms

```python
# Get antonyms (from lemmas, not synsets!)
synset = wn.synset('good.a.01')
lemma = synset.lemmas()[0]  # Get first lemma

antonyms = lemma.antonyms()
if antonyms:
    print('Antonym of good:', antonyms[0].name())
else:
    print('No antonyms found')
```

```python
# Find ALL antonyms for a word
def get_antonyms(word):
    antonyms = []
    for synset in wn.synsets(word):
        for lemma in synset.lemmas():
            for antonym in lemma.antonyms():
                antonyms.append(antonym.name())
    return list(set(antonyms))

print('Antonyms of happy:', get_antonyms('happy'))
print('Antonyms of good:', get_antonyms('good'))
```

## PATTERN 6: WordNet - Hypernyms/Hyponyms

```python
# Hypernyms (more general terms - "is a" relationship going UP)
dog = wn.synset('dog.n.01')
print('Hypernyms of dog:', dog.hypernyms())
```

```python
# Hyponyms (more specific terms - "is a" relationship going DOWN)
print('Hyponyms of dog:', dog.hyponyms()[:5])  # First 5
```

## COMPLETE EXAMPLE: Synonyms & Antonyms

```python
def get_synonyms_antonyms(word):
    """Get all synonyms and antonyms for a word"""
    synonyms = []
    antonyms = []
    
    for synset in wn.synsets(word):
        for lemma in synset.lemmas():
            synonyms.append(lemma.name())
            for ant in lemma.antonyms():
                antonyms.append(ant.name())
    
    return list(set(synonyms)), list(set(antonyms))

# Usage
syns, ants = get_synonyms_antonyms('good')
print('Synonyms:', syns)
print('Antonyms:', ants)
```
