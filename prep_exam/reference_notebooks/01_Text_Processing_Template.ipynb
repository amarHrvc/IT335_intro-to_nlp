# 01 - Text Processing Template
**MIDTERM: Text cleaning, tokenization, regex, frequency distributions**

```python
# === IMPORTS ===
import re
import string
import nltk
from nltk import FreqDist
import matplotlib.pyplot as plt
```

---
## PATTERN 1: Read File

```python
# Read entire file content
with open('filename.txt', 'r') as f:
    text = f.read()
print(text)
```

---
## PATTERN 2: Remove Punctuation

```python
# Remove ALL punctuation
clean_text = text.translate(str.maketrans('', '', string.punctuation))
print(clean_text)
```

```python
# Remove punctuation EXCEPT @ (for email extraction)
delete_chars = string.punctuation.replace('@', '')
clean_except_at = text.translate(str.maketrans('', '', delete_chars))
print(clean_except_at)
```

---
## PATTERN 3: Convert to Lowercase

```python
text_lower = clean_text.lower()
print(text_lower)
```

---
## PATTERN 4: Extract Emails with Regex

```python
# Email pattern from lectures
email_pattern = r'[\w\.]+@[\w\.]+'

emails = re.findall(email_pattern, text)
print('Emails found:', emails)
```

---
## PATTERN 5: Tokenization

```python
# Word tokenization
tokens = nltk.word_tokenize(text_lower)
print('Tokens:', tokens)
print('Total tokens:', len(tokens))
```

```python
# Keep only alphabetic tokens (remove numbers, punctuation)
alpha_tokens = [token for token in tokens if token.isalpha()]
print('Alphabetic tokens:', len(alpha_tokens))
```

---
## PATTERN 6: Frequency Distribution

```python
# Create frequency distribution
fdist = FreqDist(alpha_tokens)

# Top N most common
top_10 = fdist.most_common(10)
print('Top 10 most common:', top_10)

# Frequency of specific word
print('Frequency of "the":', fdist['the'])

# Total tokens
print('Total tokens (N):', fdist.N())
```

```python
# Plot frequency distribution
fdist.plot(20)
plt.show()
```

---
## PATTERN 7: Conditional Frequency Distribution

```python
# CFD: Word length by starting letter (for letters a, b, c)
cfd = nltk.ConditionalFreqDist(
    (word[0], len(word))
    for word in alpha_tokens 
    if word[0] in 'abc'
)

# Plot the CFD
cfd.plot()
plt.show()
```

```python
# Tabulate CFD
cfd.tabulate()
```

---
## COMPLETE WORKFLOW: Clean + Tokenize + Frequency

```python
def process_text(filename):
    """Complete text processing workflow"""
    # 1. Read file
    with open(filename, 'r') as f:
        text = f.read()
    
    # 2. Extract emails BEFORE cleaning
    email_pattern = r'[\w\.]+@[\w\.]+'
    emails = re.findall(email_pattern, text)
    
    # 3. Remove punctuation
    clean = text.translate(str.maketrans('', '', string.punctuation))
    
    # 4. Lowercase
    clean = clean.lower()
    
    # 5. Tokenize
    tokens = nltk.word_tokenize(clean)
    
    # 6. Keep only alphabetic
    alpha_tokens = [t for t in tokens if t.isalpha()]
    
    # 7. Frequency distribution
    fdist = FreqDist(alpha_tokens)
    
    return clean, alpha_tokens, emails, fdist

# Usage:
# clean_text, tokens, emails, fdist = process_text('task1.txt')
# print('Top 10:', fdist.most_common(10))
```

## REGEX QUICK REFERENCE

```python
# Common regex patterns
sample = "Order #12345 costs $99.99 email: test@example.com"

# Find all digits
digits = re.findall(r'\d+', sample)
print('Digits:', digits)  # ['12345', '99', '99']

# Find all words
words = re.findall(r'\w+', sample)
print('Words:', words)  # ['Order', '12345', 'costs', '99', '99', ...]

# Find emails
emails = re.findall(r'[\w\.]+@[\w\.]+', sample)
print('Emails:', emails)  # ['test@example.com']
```
